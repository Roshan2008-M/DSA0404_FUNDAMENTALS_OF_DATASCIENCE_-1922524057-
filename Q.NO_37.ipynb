```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
from google.colab import files

# Upload dataset
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read Excel dataset
df = pd.read_excel(filename)

# Calculate correlation
r, p = pearsonr(df['Study_Time_Hours'], df['Exam_Score'])

print("CORRELATION ANALYSIS")
print("--------------------")
print("Average Study Time:", round(df['Study_Time_Hours'].mean(), 2))
print("Average Exam Score:", round(df['Exam_Score'].mean(), 2))
print("Correlation Coefficient:", round(r, 3))

if r > 0:
    print("Positive Correlation")
elif r < 0:
    print("Negative Correlation")
else:
    print("No Linear Correlation")

# Scatter Plot
plt.figure(figsize=(7, 5))
plt.scatter(df['Study_Time_Hours'], df['Exam_Score'])
plt.xlabel("Study Time (Hours)")
plt.ylabel("Exam Score")
plt.title("Study Time vs Exam Score")
plt.grid()
plt.show()

# Regression Plot
sns.regplot(x='Study_Time_Hours', y='Exam_Score', data=df)
plt.title("Regression Plot")
plt.show()

# Line Plot
sns.lineplot(x='Study_Time_Hours', y='Exam_Score', data=df)
plt.title("Study Time and Exam Score Line Plot")
plt.show()

# Histogram
plt.hist(df['Exam_Score'], bins=10, edgecolor='black')
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")
plt.show()
```
