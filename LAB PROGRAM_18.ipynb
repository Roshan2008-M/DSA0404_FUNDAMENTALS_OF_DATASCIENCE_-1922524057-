```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from google.colab import files

# Upload dataset
uploaded = files.upload()

# Read Excel dataset
file_name = list(uploaded.keys())[0]
df = pd.read_excel(file_name)

# Select Age and BodyFat columns
data = df[['Age', 'BodyFat']]

# Calculate mean, median and standard deviation
print("Statistical Analysis")
print("----------------------------")

print("\nMean:")
print(data.mean())

print("\nMedian:")
print(data.median())

print("\nStandard Deviation:")
print(data.std())

# Boxplots
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.boxplot(y=data['Age'])
plt.title("Boxplot of Age")

plt.subplot(1, 2, 2)
sns.boxplot(y=data['BodyFat'])
plt.title("Boxplot of Body Fat")

plt.tight_layout()
plt.show()

# Scatter plot
plt.figure(figsize=(6, 5))
plt.scatter(data['Age'], data['BodyFat'])
plt.xlabel("Age")
plt.ylabel("Body Fat (%)")
plt.title("Scatter Plot of Age vs Body Fat")
plt.show()

# Q-Q plots
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
stats.probplot(data['Age'], dist="norm", plot=plt)
plt.title("Q-Q Plot of Age")

plt.subplot(1, 2, 2)
stats.probplot(data['BodyFat'], dist="norm", plot=plt)
plt.title("Q-Q Plot of Body Fat")

plt.tight_layout()
plt.show()
```
