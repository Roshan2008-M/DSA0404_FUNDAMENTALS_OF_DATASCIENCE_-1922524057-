```python

import pandas as pd
from google.colab import files

# Upload dataset
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read Excel or CSV file
if filename.endswith('.csv'):
    df = pd.read_csv(filename)
else:
    df = pd.read_excel(filename)

# Select temperature columns
cities = df.columns[1:]

# Calculate mean and standard deviation
mean_temp = df[cities].mean()
std_temp = df[cities].std()

# Calculate temperature range
temp_range = df[cities].max() - df[cities].min()

# Display results
print("TEMPERATURE VARIABILITY ANALYSIS")
print("--------------------------------")

print("\nMean Temperature:")
print(mean_temp.round(2))

print("\nStandard Deviation:")
print(std_temp.round(2))

print("\nTemperature Range:")
print(temp_range.round(2))

# City with highest temperature range
highest_range_city = temp_range.idxmax()
print("\nCity with Highest Temperature Range:")
print(highest_range_city.replace('_Temp_C', ''))

# City with most consistent temperature
consistent_city = std_temp.idxmin()
print("\nCity with Most Consistent Temperature:")
print(consistent_city.replace('_Temp_C', ''))
```
