```python
import numpy as np

def sales_analysis():
    sales_data = np.array([10000, 12000, 15000, 20000])

    total_sales = np.sum(sales_data)

    percentage_increase = ((sales_data[3] - sales_data[0]) / sales_data[0]) * 100

    print("Total Sales:", total_sales)
    print("Percentage Increase:", percentage_increase, "%")

sales_analysis()
```

```python
import numpy as np

def fuel_analysis():
    fuel_efficiency = np.array([20, 25, 30, 35, 40])

    average = np.mean(fuel_efficiency)

    percentage_improvement = ((fuel_efficiency[4] - fuel_efficiency[0]) / fuel_efficiency[0]) * 100

    print("Average Fuel Efficiency:", average, "MPG")
    print("Percentage Improvement:", percentage_improvement, "%")

fuel_analysis()
```
