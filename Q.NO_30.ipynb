```python

# CART for Car Price Prediction

import pandas as pd
from google.colab import files
from sklearn.tree import DecisionTreeRegressor
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

# Step 1: Upload Dataset
print("Upload your Car Dataset")
uploaded = files.upload()

file_name = next(iter(uploaded))
data = pd.read_excel(file_name)

print("\nDataset Loaded Successfully!")
print(data.head())

# Step 2: Select Features and Target
X = data[['Brand', 'Engine_Type', 'Age_Years', 'Mileage_Km']]
y = data['Selling_Price_USD']

# Step 3: Convert Categorical Features
preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(handle_unknown='ignore'),
     ['Brand', 'Engine_Type']),
    ('num', 'passthrough', ['Age_Years', 'Mileage_Km'])
])

# Step 4: Create CART Model
model = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', DecisionTreeRegressor(random_state=42))
])

# Step 5: Train Model
model.fit(X, y)

print("\nCART Model Trained Successfully!")

# Step 6: Get New Car Details
print("\nEnter New Car Details")

brand = input("Enter Car Brand: ")
engine = input("Enter Engine Type: ")
age = int(input("Enter Car Age (Years): "))
mileage = float(input("Enter Mileage (Km): "))

# Step 7: Predict Car Price
new_car = pd.DataFrame([[
    brand, engine, age, mileage
]], columns=X.columns)

predicted_price = model.predict(new_car)

print("\nPredicted Car Price: $",
      round(predicted_price[0], 2))

# Step 8: Display Decision Path
transformed_data = model.named_steps['preprocessor'].transform(new_car)

tree_model = model.named_steps['regressor']
tree = tree_model.tree_

feature_names = model.named_steps['preprocessor'].get_feature_names_out()

path = tree_model.decision_path(transformed_data)

print("\nDecision Path:")

for node in path.indices:
    if tree.feature[node] != -2:
        feature = feature_names[tree.feature[node]]
        threshold = tree.threshold[node]
        value = transformed_data[0, tree.feature[node]]

        if value <= threshold:
            print(f"{feature} <= {threshold:.2f}")
        else:
            print(f"{feature} > {threshold:.2f}")

print("\nFinal Predicted Price: $",
      round(predicted_price[0], 2))
```
