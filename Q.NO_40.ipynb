```python

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Upload Excel dataset
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read Excel file
df = pd.read_excel(filename)

# Create and save CSV file
df.to_csv('soccer_players.csv', index=False)

# Read data from CSV file
data = pd.read_csv('soccer_players.csv')

# Display dataset
print("SOCCER PLAYER DATA")
print(data)

# Top 5 players by goals
print("\nTOP 5 PLAYERS BY GOALS")
print(data.nlargest(5, 'Goals_Scored')[
    ['Player_Name', 'Goals_Scored']
])

# Top 5 players by salary
print("\nTOP 5 PLAYERS BY SALARY")
print(data.nlargest(5, 'Weekly_Salary_USD')[
    ['Player_Name', 'Weekly_Salary_USD']
])

# Calculate average age
avg_age = data['Age'].mean()
print("\nAverage Age:", round(avg_age, 2))

# Players above average age
print("\nPLAYERS ABOVE AVERAGE AGE")
print(data[data['Age'] > avg_age][
    ['Player_Name', 'Age']
])

# Count players by position
position_count = data['Position'].value_counts()

print("\nPLAYERS BY POSITION")
print(position_count)

# Bar chart
position_count.plot(kind='bar', figsize=(8, 5),
                    color='skyblue', edgecolor='black')

plt.xlabel('Player Position')
plt.ylabel('Number of Players')
plt.title('Distribution of Soccer Players by Position')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()

# Download CSV file
files.download('soccer_players.csv')
```
