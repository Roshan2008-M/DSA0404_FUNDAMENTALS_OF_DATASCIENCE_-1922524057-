```python
from google.colab import files
uploaded = files.upload()
import pandas as pd

# Read dataset
df = pd.read_excel("Sales (1) (1).xlsx")

# Select first 5 products as a sample purchase
prices = df["Price"].head(5).tolist()
quantities = df["Quantity_Sold"].head(5).tolist()
discounts = df["Discount"].head(5).tolist()

# Calculate total amount
total = 0

for i in range(len(prices)):
    total += prices[i] * quantities[i]

    # Calculate discount
    discount = 0

    for i in range(len(prices)):
        discount += prices[i] * quantities[i] * discounts[i]

        # Amount after discount
        amount = total - discount

        # Calculate tax
        tax_rate = 5
        tax = amount * tax_rate / 100

        # Calculate final cost
        final_cost = amount + tax

        # Display results
        print("Total Amount:", round(total, 2))
        print("Discount:", round(discount, 2))
        print("Amount After Discount:", round(amount, 2))
        print("Tax:", round(tax, 2))
        print("Final Cost:", round(final_cost, 2))
```

```python
from google.colab import files
uploaded = files.upload()
import pandas as pd

# Read the dynamically uploaded file
file_name = next(iter(uploaded))
df = pd.read_excel(file_name)

# Convert order date into datetime format
df["order_date"] = pd.to_datetime(df["order_date"])

# 1. Total number of orders made by each customer
customer_orders = df.groupby("customer_name")["order_id"].nunique()

print("Total Orders Made by Each Customer:")
print(customer_orders.head(10))

# 2. Average order quantity for each product
average_quantity = df.groupby("product_name")["quantity"].mean()

print("\nAverage Order Quantity for Each Product:")
print(average_quantity.head(10))

# 3. Earliest and latest order dates
earliest_date = df["order_date"].min()
latest_date = df["order_date"].max()

print("\nEarliest Order Date:", earliest_date)
print("Latest Order Date:", latest_date)
```

```python
import pandas as pd
from google.colab import files

# Upload the dataset
uploaded = files.upload()

# Read the Excel file dynamically
file_name = next(iter(uploaded))
df = pd.read_excel(file_name)

# Convert order_date into date format
df["order_date"] = pd.to_datetime(
    df["order_date"], errors="coerce", format="mixed"
)

# Find the latest month in the dataset
latest_month = df["order_date"].max().to_period("M")

# Filter the latest month's data
monthly_data = df[
    df["order_date"].dt.to_period("M") == latest_month
]

# Find the top 5 products
top_5 = (
    monthly_data.groupby("product_name")["quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("Top 5 Products Sold:")
print(top_5)
```

```python
import re
from collections import Counter
from google.colab import files

# Upload the text file
uploaded = files.upload()

# Read the file
with open(next(iter(uploaded)), "r") as file:
    text = file.read()

    # Extract words and calculate frequency
    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
    frequency = Counter(words)

    # Display word frequency
    print("Word Frequency Distribution:")
    for word, count in frequency.items():
        print(word, ":", count)
```
