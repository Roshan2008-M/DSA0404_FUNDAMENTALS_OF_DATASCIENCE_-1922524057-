```python
# Seaborn Important Graphs and Functions in One Program
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# -----------------------------------------
# 1. Create Sample Dataset
# -----------------------------------------

data = {
    "Student": ["Arun", "Bala", "Chitra", "Divya", "Ezhil", "Farhan", "Geetha", "Hari"],
    "Department": ["CSE", "AI", "CSE", "AI", "IT", "IT", "CSE", "AI"],
    "Gender": ["Male", "Male", "Female", "Female", "Male", "Male", "Female", "Male"],
    "Study_Hours": [2, 4, 3, 5, 6, 2, 4, 5],
    "Marks": [50, 70, 65, 85, 90, 55, 75, 80],
    "Attendance": [70, 85, 80, 95, 90, 65, 88, 92]
}
df = pd.DataFrame(data)

print("Sample DataFrame:")
print(df)

# -----------------------------------------
# 2. Set Seaborn Theme
# -----------------------------------------

sns.set_theme(style="whitegrid")

# -----------------------------------------
# 3. Line Plot
# -----------------------------------------

plt.figure(figsize=(6, 4))
sns.lineplot(data=df, x="Study_Hours", y="Marks", marker="o")
plt.title("Line Plot: Study Hours vs Marks")
plt.show()

# -----------------------------------------
# 4. Scatter Plot
# -----------------------------------------

plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x="Study_Hours", y="Marks", hue="Department")
plt.title("Scatter Plot: Study Hours vs Marks")
plt.show()

# -----------------------------------------
# 5. Bar Plot
# -----------------------------------------

plt.figure(figsize=(6, 4))
sns.barplot(data=df, x="Department", y="Marks")
plt.title("Bar Plot: Average Marks by Department")
plt.show()

# -----------------------------------------
# 6. Count Plot
# -----------------------------------------

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Department")
plt.title("Count Plot: Number of Students by Department")
plt.show()

# -----------------------------------------
# 7. Box Plot
# -----------------------------------------

plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x="Department", y="Marks")
plt.title("Box Plot: Marks Distribution by Department")
plt.show()

# -----------------------------------------
# 8. Violin Plot
# -----------------------------------------
plt.figure(figsize=(6, 4))
sns.violinplot(data=df, x="Department", y="Marks")
plt.title("Violin Plot: Marks Distribution by Department")
plt.show()

# -----------------------------------------
# 9. Histogram
# -----------------------------------------
plt.figure(figsize=(6, 4))
sns.histplot(data=df, x="Marks", bins=5, kde=True)
plt.title("Histogram: Marks Distribution")
plt.show()

# -----------------------------------------
# 10. KDE Plot
# -----------------------------------------

plt.figure(figsize=(6, 4))
sns.kdeplot(data=df, x="Marks", fill=True)
plt.title("KDE Plot: Marks Density")
plt.show()

# -----------------------------------------
# 11. Pair Plot
# -----------------------------------------

sns.pairplot(df[["Study_Hours", "Marks", "Attendance"]])
plt.suptitle("Pair Plot: Numeric Variable Relationships", y=1.02)
plt.show()

# -----------------------------------------
# 12. Heatmap
# -----------------------------------------
correlation = df[["Study_Hours", "Marks", "Attendance"]].corr()

plt.figure(figsize=(6, 4))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Heatmap: Correlation Matrix")
plt.show()

# -----------------------------------------
# 13. Regression Plot
# -----------------------------------------

plt.figure(figsize=(6, 4))
sns.regplot(data=df, x="Study_Hours", y="Marks")
plt.title("Regression Plot: Study Hours vs Marks")
plt.show()

# -----------------------------------------
# 14. Strip Plot
# -----------------------------------------

plt.figure(figsize=(6, 4))
sns.stripplot(data=df, x="Department", y="Marks")
plt.title("Strip Plot: Marks by Department")
plt.show()

# -----------------------------------------
# 15. Swarm Plot
# -----------------------------------------
plt.figure(figsize=(6, 4))
sns.swarmplot(data=df, x="Department", y="Marks")
plt.title("Swarm Plot: Marks by Department")
plt.show()

print("Seaborn Program Completed Successfully")
```
