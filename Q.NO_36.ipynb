```python

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Upload dataset
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read CSV or Excel file
if filename.endswith('.csv'):
    df = pd.read_csv(filename)
else:
    df = pd.read_excel(filename)

# Extract closing prices
prices = df['Close']

# Calculate statistical measures
mean_price = prices.mean()
variance = prices.var()
std_dev = prices.std()
minimum = prices.min()
maximum = prices.max()
price_range = maximum - minimum

# Display results
print("STOCK PRICE VARIABILITY ANALYSIS")
print("--------------------------------")
print("Mean Closing Price:", round(mean_price, 2))
print("Variance:", round(variance, 2))
print("Standard Deviation:", round(std_dev, 2))
print("Minimum Price:", minimum)
print("Maximum Price:", maximum)
print("Price Range:", round(price_range, 2))

# Display price movement
change = prices.iloc[-1] - prices.iloc[0]
print("\nPrice Movement:", round(change, 2))

if change > 0:
    print("Stock Price Trend: Increasing")
elif change < 0:
    print("Stock Price Trend: Decreasing")
else:
    print("Stock Price Trend: No Change")

# Plot closing prices
plt.figure(figsize=(10, 5))
plt.plot(df['Date'], prices, marker='o')
plt.xlabel('Date')
plt.ylabel('Closing Price')
plt.title('Stock Price Movement Analysis')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()

```
