```python

import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files

# Upload Excel dataset
uploaded = files.upload()

# Read Excel file
df = pd.read_excel("rare_elements.xlsx")

# Extract concentration values
data = df["Concentration_ppm"].dropna().to_numpy()

# Get user input
n = int(input("Enter sample size: "))
confidence = float(input("Enter confidence level (%): "))
precision = float(input("Enter desired precision: "))

# Check inputs
if n < 2 or n > len(data):
    print("Invalid sample size!")
elif not (0 < confidence < 100) or precision <= 0:
    print("Invalid confidence level or precision!")
else:
    # Select random sample
    np.random.seed(42)
    sample = np.random.choice(data, n, replace=False)

    # Calculate point estimate
    mean = np.mean(sample)
    std = np.std(sample, ddof=1)

    # Calculate confidence interval
    alpha = 1 - confidence / 100
    t_value = stats.t.ppf(1 - alpha / 2, n - 1)

    margin_error = t_value * std / np.sqrt(n)

    lower = mean - margin_error
    upper = mean + margin_error

    # Display results
    print("\n--- Rare Element Estimation ---")
    print("Total Measurements:", len(data))
    print("Sample Size:", n)
    print("Sample Mean:", round(mean, 3))
    print("Standard Deviation:", round(std, 3))
    print("Confidence Level:", confidence, "%")
    print("Confidence Interval:",
          round(lower, 3), "to", round(upper, 3))
    print("Margin of Error:", round(margin_error, 3))

    if margin_error <= precision:
        print("Desired precision is achieved.")
    else:
        print("Desired precision is not achieved.")
```
