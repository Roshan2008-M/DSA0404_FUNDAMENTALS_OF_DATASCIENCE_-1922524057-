```python
%%writefile ATMWithdrawal.java
import java.util.Scanner;

class ATMWithdrawal {
    public static void main(String[] args) {
        // Note: Interactive Scanner input might not work smoothly with standard non-interactive execution
        // You can hardcode a value or run the compiled class in a terminal
        Scanner sc = new Scanner("2000\n"); // Mocking input for non-interactive notebook execution

        int balance = 5000;
        System.out.print("Enter withdrawal amount: ");
        int amount = sc.nextInt();
        System.out.println(amount);

        try {
            if (amount > balance)
                throw new ArithmeticException("Insufficient balance");

            balance -= amount;
            System.out.println("Withdrawal successful!");
            System.out.println("Remaining balance: " + balance);
        } catch (ArithmeticException e) {
            System.out.println(e.getMessage());
        }
    }
}

```

```python
%%writefile AgeCheck.java
class InvalidAgeException extends Exception {
    InvalidAgeException(String message) {
        super(message);
    }
}

class AgeCheck {
    static void checkAge(int age) throws InvalidAgeException {
        if (age < 18)
            throw new InvalidAgeException("Age is not eligible");
        else
            System.out.println("Eligible");
    }

    public static void main(String[] args) {
        try {
            checkAge(16);
        } catch (InvalidAgeException e) {
            System.out.println(e.getMessage());
        }
    }
}
```

```python
!javac AgeCheck.java && java AgeCheck
```

```python
%%writefile SalesReport.java
import java.io.*;

class SalesReport {
    public static void main(String[] args) {
        try (BufferedReader br = new BufferedReader(new FileReader("sales.txt"))) {
            String line;
            while ((line = br.readLine()) != null) {
                System.out.println(line);
            }
        } catch (IOException e) {
            System.out.println("File error: " + e.getMessage());
        }
    }
}
```

```python
%%writefile ThreadDemo.java
class MyThread extends Thread {
    public void run() {
        for (int i = 1; i <= 5; i++) {
            System.out.println("Thread: " + i);
        }
    }
}

class ThreadDemo {
    public static void main(String[] args) {
        MyThread t = new MyThread();
        t.start();

        for (int i = 1; i <= 5; i++) {
            System.out.println("Main: " + i);
        }
    }
}
```

```python
%%writefile BookingDemo.java
class SeatBooking {
    int seats = 1;

    synchronized void bookSeat(String name) {
        if (seats > 0) {
            System.out.println(name + " booked a seat");
            seats--;
        } else {
            System.out.println(name + " no seats available");
        }
    }
}

class BookingDemo {
    public static void main(String[] args) {
        SeatBooking b = new SeatBooking();

        Thread t1 = new Thread(() -> b.bookSeat("Student 1"));
        Thread t2 = new Thread(() -> b.bookSeat("Student 2"));

        t1.start();
        t2.start();
    }
}
```

```python
!javac BookingDemo.java && java BookingDemo
```

```python
%%writefile CommunicationDemo.java
class TokenCounter {
    int token = 0;

    synchronized void produce() {
        token++;
        System.out.println("Produced: " + token);
        notify();
    }

    synchronized void consume() {
        try {
            wait();
            System.out.println("Consumed: " + token);
        } catch (InterruptedException e) {
            System.out.println(e);
        }
    }
}

class CommunicationDemo {
    public static void main(String[] args) {
        TokenCounter c = new TokenCounter();

        Thread t1 = new Thread(() -> c.consume());
        Thread t2 = new Thread(() -> c.produce());

        t1.start();
        t2.start();
    }
}
```

```python
!javac CommunicationDemo.java && java CommunicationDemo
```

```python
%%writefile ResultDemo.java
import java.io.*;

class ResultProcessor extends Thread {
    String file;

    ResultProcessor(String file) {
        this.file = file;
    }

    public void run() {
        try (BufferedReader br = new BufferedReader(new FileReader(file))) {
            String line;
            int total = 0;

            while ((line = br.readLine()) != null) {
                total += Integer.parseInt(line);
            }

            System.out.println(file + " Total: " + total);
        } catch (Exception e) {
            System.out.println("Error: " + e.getMessage());
        }
    }
}

class ResultDemo {
    public static void main(String[] args) {
        ResultProcessor t1 = new ResultProcessor("marks1.txt");
        ResultProcessor t2 = new ResultProcessor("marks2.txt");
        ResultProcessor t3 = new ResultProcessor("marks3.txt");

        t1.start();
        t2.start();
        t3.start();
    }
}
```

```python
%%bash
# Create the sample marks files
echo -e "85\n90\n78" > marks1.txt
echo -e "92\n88\n95" > marks2.txt
echo -e "70\n82\n80" > marks3.txt

# Compile and run the Java program
javac ResultDemo.java && java ResultDemo
```
